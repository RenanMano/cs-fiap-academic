# **Atividade - Motores - SERS**

**Objetivo:**

Com base nos dados de placa de motores de indução (PU, n, FP), calcular as potências ativa, reativa e aparente (P, Q e S).

*1. Inserção dos dados do motor*

```python
# Potência Útil (Mecânica) ----> PU ----> em Watts
PU = int(input('Insira o valor da potência em Watts: '))
# Rendimento (de 0 a 1) ----> n
n = float(input('Insira o valor do rendimento (de 0 a 1): '))
# Fator de potência (de 0 a 1) ----> FP (cosPhi)
FP = float(input('Insira o fator de potência (de 0 a 1): '))
```

```python
PU, n, FP
```

*2. Cálculo das potências ativa (P), aparente (S) e reativa (Q)*

P = PU / n

S = P / FP

Q ----> teorema de Pitágoras

```python
# Potência Ativa (P)
P = PU / n
# Potência Aparente (S)
S = P / FP
# Potência Reativa (Q)
# Exponenciação ---> **
# Raiz quadrada ---> ** 0.5
Q = (S**2 - P**2)**0.5
```

```python
P, S, Q
```

*3. Apresentação dos resultados*

```python
print(f'A potência ativa será de {int(P)} W')
print(f'A potência aparente será de {int(S)} VA')
print(f'A potência reativa será de {int(Q)} VAr')

# Exemplo para limitar o resultado em duas casas decimais
# print(f'A potência ativa será de {P:.2f} W')
```
