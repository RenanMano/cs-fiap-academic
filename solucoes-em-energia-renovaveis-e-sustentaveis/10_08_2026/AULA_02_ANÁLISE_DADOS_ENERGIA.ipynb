```python
# Calcular a demanda máxima (Potência Ativa, em kW)
PMAX = dados['Potencia_Ativa'].max()
print(f'A demanda máxima de potência é {PMAX:.2f} kW')
```

```python
# Calcular limiar de 70% da demanda máxima (PMAX)
P70 = 0.7 * PMAX
print(f'O limiar de 70% da demanda máxima é {P70:.2f} kW')
```

```python
# Quantos consumidores estão com demanda acima de 70%
# Criar um dataframe 'df4' somente com as colunas de tensão, corrente e potências
# e registros de consumidores com demanda energética acima de 70% de PMAX
```

```python
dados.columns
```

```python
df4 = dados[['Tensao', 'Corrente', 'Potencia_Ativa', 'Potencia_Reativa']]
df4.head()
```

```python
df4.shape[0]
```

```python
df5 = df4[df4['Potencia_Ativa'] > P70]
df5.head()
```

```python
df5.shape[0]
```

```python
print(f'Foram encontrados {df5.shape[0]} consumidores com demanda energética acima de 70% de PMAX')
```
