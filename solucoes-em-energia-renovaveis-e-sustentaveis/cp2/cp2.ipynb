```python
dados = pd.read_csv('https://raw.githubusercontent.com/RenanMano/cp02-solucoes-em-energia-renovaveis-e-sustentaveis/refs/heads/main/Data_for_UCI_named.csv')
dados.head()
```

freatures

```python
# freatures
x = dados.drop(columns=['stab', 'stabf'], axis=1)
x.head()
```

```python
# target (stable ou unstable)
y = dados['stabf']
y.head()
```

```python
# quais classes temos na coluna alvo?
dados['stabf'].unique()
```

```python
# quantos registros temos na coluna alvo?
dados['stabf'].value_counts()
```

```python
# separação de dados de treino e teste
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
x.shape
```

```python
x_train.shape
```

```python
x_test.shape
```

instanciando a classe logisticRegression


```python
modelo = LogisticRegression()
```

treinamento do modelo

```python
modelo.fit(x_train, y_train)
```

gerando o array de probabilidades

```python
probabilidades = modelo.predict_proba(x_test)
probabilidades[:50]
```

gerando o array de previsões

```python
y_predict = modelo.predict(x_test)
y_predict[:50]
```

```python
# acurácia
acuracia = 100 * accuracy_score(y_test, y_predict)
print(f'a acurácia do modelo baseado em regressão logistica é: {acuracia:.2f}%')
```
