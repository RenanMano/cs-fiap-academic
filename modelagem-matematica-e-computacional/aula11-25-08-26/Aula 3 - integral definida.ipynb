```python
import numpy as np

x = np.linspace(0, 8, 1000000)
def f(x):
    return -(x**2)/4 + 2*x
y = f(x)
```

```python
import matplotlib.pyplot as plt

plt.plot(x, y)
plt.axhline(linewidth=0.5, color='black')
plt.axvline(linewidth=0.5, color='black')

plt.show()
```

```python
np.trapz(y, x)
```

```python
base = 0.001

```

```python
intervalos = np.arange(0, 8, base)
med = np.arange((0 + base)/2, 8, base)
altura = f(med)
area = base*altura 
soma = area.sum()

soma
```
