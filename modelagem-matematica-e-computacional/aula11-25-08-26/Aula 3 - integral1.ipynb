```python
import sympy as sp
```

```python
x = sp.symbols('x')
```

```python
f =  x
f
```

```python
sp.integrate(f, x)
```

```python
sp.integrate(f, (x,-2,2))
```

```python
sp.integrate(f, (x,-2,0))
```

```python
sp.integrate(f, (x,0,2))
```

```python
import numpy as np

X = np.arange(-2, 2, 0.01)
y = sp.lambdify(x, f, 'numpy')
F = y(X)
```

```python
import matplotlib.pyplot as plt

plt.plot(X, F)
plt.axhline(color='black', linewidth=0.5)
plt.axvline(color='black', linewidth=0.5)
plt.fill_between(X, F, 0, where=((F>=0) & (X<=2)), color='blue', alpha=0.3)
plt.fill_between(X, F, 0, where=((F<=0) & (X<0)), color='red', alpha=0.3)

plt.show()

```
