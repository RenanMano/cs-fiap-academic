## **VETORES**

```python
import numpy as np
```

```python
v = np.array([2, 3])
w = np.array([1, -1])

print(v)
print(w)
```

```python
w[0]
```

```python
w[1]
```

```python
w[2]
```

```python
import matplotlib.pyplot as plt
```

```python
plt.quiver(0,0,v[0],v[1], angles='xy', scale_units='xy', scale=1)
plt.quiver(0,0,w[0],w[1], angles='xy', scale_units='xy', scale=1)

plt.text(v[0], v[1], 'v', fontsize=12)
plt.text(w[0], w[1], 'w', fontsize=12)

plt.xlim(-5,5)
plt.ylim(-5,5)

plt.axhline(0)
plt.axvline(0)

plt.grid()
plt.show()
```

```python
2*v
# v = [2, 3]
```

```python
w/2
# w = [1, -1]
```

```python
print(v+w) # [(2+1),(3-1)] = [3,2]
print(v)
print(w)
```

```python
v-w # [(2-1),(3-(-1))] = [1,4]
```

```python
np.dot(v,w)

# Produto escalar / Dot Product
# v.w = (2)(1)+(3)(-1) = -1

```

```python
v@w
```

```python
np.linalg.vecdot(v,w)
```

```python
E = np.array([[1, 2],
              [3, 4]])

F = np.array([[5, 6],
              [7, 8]])

print(E + F)
```

```python
print(3 * E)
```

```python
print(np.matmul(E, F))

# 1ª linha de E pela 1ª coluna de F
  # (1)5 + (2)7 = 19
# 1ª linha de E pela 2ª coluna de F
  # (1)6 + (2)8 = 22
```

```python
print(np.matmul(F, E))
```

# **SISTEMAS LINEARES**

```python
import numpy as np
```

```python
A = np.array([
    [1, 1, 1],
    [2, -1, 1],
    [1, 2, -1]
])

b = np.array([6, 3, 3])

# b = np.array([[6],[3],[3]])
```

```python
A
```

```python
b
```

```python
np.shape(b)
```

```python
# Ax = b

x = np.linalg.solve(A, b)
print(x)
```
