```python
import sympy as sp
```

```python
x = sp.symbols('x')
```

```python
f = sp.sin(x)
```

```python
f
```

```python
sp.integrate(f,x)
```

```python
import numpy as np

sp.integrate(f, (x, 0, np.pi))
```

```python
x1 = np.arange(0, np.pi, 0.00001)
y1 = np.sin(x1)
```

```python
np.trapz(y1, x1)
```

```python
import matplotlib.pyplot as plt
```

```python
X = np.arange(np.pi*(-2), np.pi*(2), 0.1)
y = sp.lambdify(x, f, 'numpy')
F = y(X)
```

```python
plt.plot(X, F, label='sen(x)', color='red', linewidth=0.75)
plt.axvline(color='black', linewidth=0.5)
plt.axhline(color='black', linewidth=0.5)

plt.xticks(
    [-2*np.pi, -np.pi, 0, np.pi, 2*np.pi],
    [r'$-2\pi$', r'$-\pi$', r'$0$', r'$\pi$', r'$2\pi$']
)

plt.fill_between(X, F, where=((X>0) & (F>0)), color='blue', alpha=0.3)

plt.legend()
plt.show()
```
