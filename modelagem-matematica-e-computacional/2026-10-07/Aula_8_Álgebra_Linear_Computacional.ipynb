# **SISTEMAS LINEARES**

```python
import numpy as np
```

```python
A = np.array([
    [1, 1, 1],
    [1, -1, 1],
    [2, 1, -1]
])

b = np.array([[6],[2],[7]])

```

```python
A
```

```python
np.shape(A)
```

```python
b
```

```python
np.shape(b)
```

```python
# Ax = b

x = np.linalg.solve(A, b)
print(x)
```

```python
Aumentada = np.hstack((A, b))
Aumentada
```

```python
posto_A = np.linalg.matrix_rank(A)
posto_aug = np.linalg.matrix_rank(Aumentada)
n_variaveis = A.shape[1]

print("Posto de A:", posto_A)
print("Posto da aumentada:", posto_aug)
print("Número de variáveis:", n_variaveis)

if posto_A == n_variaveis:
    print("Sistema SPD")
elif posto_A < n_variaveis:
    print("Sistema SPI")
else:
    print("Sistema SI")
```

## **Escalonamento**

```python
import sympy as sp

A = sp.Matrix([
    [1, 1, 1],
    [1, -1, 1],
    [2, 1, -1]
])

b = sp.Matrix([
    [6],
    [2],
    [7]
])

Expandida = A.row_join(b)  # Matriz aumentada

A_esc, pivots = Expandida.rref()   # Matriz escalonada por Gauss-Jordan e pivôs
```

**Sistema Possível e Determinado (SPD)**

```python
import sympy as sp

A = sp.Matrix([
    [1, 1, 1],
    [2, -1, 1],
    [1, 2, -1]
])

b = sp.Matrix([
    [6],
    [2],
    [7]
])

M = A.row_join(b)
M_rref, pivots = M.rref()

print("É um sistema SPD?")
print("posto(A) =", A.rank())
print("posto([A|b]) =", M.rank())
print("número de variáveis =", A.shape[1])
sp.pprint(M_rref)
```

**Sistema Impossível (SI)**

# **DETERMINANTES**

```python
print(A)
print(A.shape)
```

```python
print(B)
print(B.shape)
```

```python
print(b)
print(b.shape)
```

```python
det_A = np.linalg.det(A)
det_B = np.linalg.det(B)

print("det(A) =", det_A)
print("det(B) =", det_B)
```

# **MATRIZ INVERSA**

```python
A = np.array([[2, 1],
              [3, 4]])

A_inv = np.linalg.inv(A)

print("A inversa é:")
print(A_inv)
```

```python
print("A @ A_inv =")
print(A @ A_inv)

print("A_inv @ A =")
print(A_inv @ A)
```

```python
b_inv = np.linalg.inv(b)  # Uma matriz que não for quadrada não possui inversa
```

```python
C = np.array([[2, 2],
              [3, 3]])
```

```python
det_C = np.linalg.det(C)
det_C
```

```python
C_inv = np.linalg.inv(C)
```

**Matriz Inversa e Sistema Linear**
